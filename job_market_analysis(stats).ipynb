```python
#application of inferenctial stats
import pandas as pd
import numpy as np

df = pd.read_csv("ai_jobs.csv")
sample = df.sample(n=500, random_state=42)

average_salary_max = sample["salary_max_usd"].mean()
average_salary_min = sample["salary_min_usd"].mean()
average_salary = (average_salary_max + average_salary_min) / 2

print("Estimated Average Salary:", average_salary)
```

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("ai_jobs.csv")
salary = df["salary_min_usd"]

ci = stats.t.interval(
    confidence=0.95,
    df=len(salary)-1,
    loc=salary.mean(),
    scale=stats.sem(salary)
)

print(ci)
```

```python
df['salary_in_usd'] = (df['salary_min_usd'] + df['salary_max_usd']) / 2
display(df.head())
```

```python
import pandas as pd
from scipy import stats

salary = df["salary_in_usd"]

mean = salary.mean()

se = stats.sem(salary)

t_value = stats.t.ppf(0.975, df=len(salary)-1)

moe = t_value * se

print("Sample Mean:", mean)
print("Standard Error:", se)
print("Margin of Error:", moe)
```

```python
import numpy as np
from scipy.stats import ttest_1samp

salary = np.array([92000, 95000, 91000, 93000, 96000])

t_stat, p_value = ttest_1samp(salary, popmean=90000)

print("T Statistic:", t_stat)
print("P-value:", p_value)
#got p values as 0.02 so reject the null hypothesis because we know that p>0.05 t0 not reject the h0
```

```python
import numpy as np
from scipy.stats import ttest_ind

remote = np.array([95000,98000,102000,97000,99000])
onsite = np.array([85000,87000,90000,88000,86000])
t_stat, p_value = ttest_ind(remote, onsite)
print("P-value:", p_value)
```

```python
import numpy as np
from scipy.stats import ttest_ind
model_A = np.array([89,90,88,89,90,91,89,88,90,89])
model_B = np.array([91,92,90,91,92,93,91,90,92,91])
t_stat, p_value = ttest_ind(model_A, model_B)

print("P-value:", p_value)
#also here p value is < 0.05 so we reject h0
```

```python
import numpy as np
from scipy.stats import ttest_ind

model_A = np.array([89,90,88,89,90,91,89])

model_B = np.array([91,92,90,91,92,93,91])
t_stat, p_value = ttest_ind(model_A, model_B)
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Reject H₀")
else:
    print("Fail to Reject H₀")
```
